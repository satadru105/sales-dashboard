# Interactive Sales Dashboard
This notebook loads the `sales_data (3).csv` dataset, performs statistical analysis, and generates both static (Seaborn) and interactive (Plotly) visualizations to meet the Week 6 project requirements.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
import warnings
warnings.filterwarnings('ignore')
```

## 1. Data Loading & Inspection
Load the dataset and check the column data types to ensure they are ready for plotting.

```python
df = pd.read_csv('sales_data (3).csv')
df['Date'] = pd.to_datetime(df['Date'])

display(df.head())
display(df.info())
```

## 2. Seaborn Statistical Visualizations
Creating a 2x2 grid containing a box plot, violin plot, bar plot, and correlation heatmap. This satisfies the requirement for multiple advanced chart types and coordinated themes.

```python
sns.set_theme(style="whitegrid", palette="muted")
fig, axes = plt.subplots(2, 2, figsize=(16, 12))
fig.suptitle('Sales Performance Dashboard', fontsize=20, fontweight='bold', y=1.02)

# 1. Box plot: Total Sales by Product
sns.boxplot(x='Product', y='Total_Sales', data=df, hue='Product', ax=axes[0, 0], palette='Set2', legend=False)
axes[0, 0].set_title('Total Sales Distribution by Product', fontsize=14)
axes[0, 0].tick_params(axis='x', rotation=45)

# 2. Violin plot: Quantity by Region
sns.violinplot(x='Region', y='Quantity', data=df, hue='Region', ax=axes[0, 1], palette='Pastel1', inner='quartile', legend=False)
axes[0, 1].set_title('Quantity Distribution by Region', fontsize=14)

# 3. Bar plot: Total Sales by Region
sales_by_region = df.groupby('Region')['Total_Sales'].sum().reset_index()
sns.barplot(x='Region', y='Total_Sales', data=sales_by_region, hue='Region', ax=axes[1, 0], palette='crest', legend=False)
axes[1, 0].set_title('Total Sales Revenue by Region', fontsize=14)

# 4. Correlation Heatmap
numeric_cols = df[['Quantity', 'Price', 'Total_Sales']]
sns.heatmap(numeric_cols.corr(), annot=True, cmap='coolwarm', ax=axes[1, 1], vmin=-1, vmax=1, center=0, square=True)
axes[1, 1].set_title('Correlation Matrix of Numerical Features', fontsize=14)

plt.tight_layout()
plt.show()
```

## 3. Interactive Plotly Visualization
Creating a dynamic scatter plot with hover information, providing an additional chart type (Bubble Scatter) that allows users to explore specific customer transactions.

```python
df_sorted = df.sort_values('Date')

fig = px.scatter(
    df_sorted, 
    x="Price", 
    y="Total_Sales", 
    size="Quantity", 
    color="Product",
    hover_name="Customer_ID",
    hover_data=["Region", "Date", "Quantity", "Price"],
    title="Interactive Sales Analysis: Price vs. Total Sales (Bubble Size = Quantity)",
    template="plotly_white"
)
fig.update_traces(marker={"opacity": 0.85, "line": {"width": 1, "color": "white"}})

fig.show()
```
